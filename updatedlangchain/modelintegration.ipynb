## Model Integration

```python
import os
from dotenv import load_dotenv
load_dotenv()


os.environ["OPENROUTER_API_KEY"] = os.environ.get("OPENROUTER_API_KEY")
os.environ["NVIDIA_API_KEY"] = os.environ.get("NVIDIA_API_KEY")
os.environ["GEMINI_API_KEY"] = os.environ.get("GEMINI_API_KEY")
```

### Google Gemini Model Integration

#### init_chat_model

```python
import os
from langchain.chat_models import init_chat_model


model = init_chat_model("google_genai:gemini-3.8-flash")
response = model.invoke("Why do parrots talk?")
response.content
```

#### langchain_google_genani

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
response = model.invoke("Why do parrots talk?")
response
```

### NVIDIA Model Integration

#### langchain_nvidia_ai_endpoints

#### init_chat_model

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("nvidia:nvidia/nemotron-3-ultra-550b-a55b")
response = model.invoke("What is GAN?")
response
```

## Streaming And Batch

### Streaming

```python
# model.stream("What is GAN?")
for chunk in model.stream("What is transformer in LLM?"):
    print(chunk.text, end="")
```

### Batch

```python
response = model.batch(["What is ANN?", "What is RNN?", "What is CNN?"])
for r in response:
    print(r)
```
